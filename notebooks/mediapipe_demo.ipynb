```python
import time
import cv2
from src.builtin.mediapipe_builtin.landmarkers_loader import LandmarkerLoader
from src.features import extract as ft_ex
from src.features.camera import open_camera
from src.features.frame import frame_padding

landmarkers = LandmarkerLoader()
cap = open_camera(0)

try:
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        frame = frame_padding(frame)
        pose_res, hand_res, timestamp = landmarkers.detect(frame)

        ft=ft_ex.extract(pose_res,hand_res)
        print(ft.shape)

        cv2.imshow("VSL Feature Debug", frame)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break
finally:
    cap.release()
    landmarkers.close()
    cv2.destroyAllWindows()
```
